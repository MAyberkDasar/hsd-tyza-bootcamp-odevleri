##### 6. indirim_orani ve musteri_puani sütunlarındaki eksik değerleri medyan ile doldurun. Doldurmadan önce ve sonra bu iki sütundaki eksik değer sayılarını karşılaştırın.

```python
df_filled = df.copy()
```

```python
df_filled[["indirim_orani", "musteri_puani"]].isna().sum()
```

```python
df_filled["indirim_orani"] = df_filled["indirim_orani"].fillna(df_filled["indirim_orani"].median())
```

```python
df_filled["musteri_puani"] = df_filled["musteri_puani"].fillna(df_filled["musteri_puani"].median())
```

```python
df_filled[["indirim_orani", "musteri_puani"]].isna().sum()
```

##### 7. odeme_turu ve musteri_tipi sütunlarındaki eksik değerleri mod ile doldurun. Daha sonra bu sütunlarda eksik değer kalıp kalmadığını kontrol edin.

```python
df_filled[["odeme_turu", "musteri_tipi"]].isna().sum()
```

```python
df_filled["odeme_turu"] = df_filled["odeme_turu"].fillna(df_filled["odeme_turu"].mode()[0])
```

```python
df_filled["musteri_tipi"] = df_filled["musteri_tipi"].fillna(df_filled["musteri_tipi"].mode()[0])
```

```python
df_filled[["odeme_turu", "musteri_tipi"]].isna().sum()
```

##### 11. kategori sütunundaki tutarsız değerleri inceleyin. Örneğin Ev Yaşam, ev yaşam, ev&yaşam gibi farklı yazımları tek biçime dönüştürün. Temizlikten sonra value_counts() çıktısını yazdırın.

##### 12. Aşağıdaki koşullara uyan mantıksal olarak hatalı kayıtları bulun: birim_fiyat <= 0, toplam_tutar <= 0, teslimat_gunu < 0, musteri_puani > 5. Bu kayıtların sayısını yazdırın ve veri setinden çıkarın.

```python
inconsistent_rows = df_inconsistent[
    (df_inconsistent["birim_fiyat"] <= 0) |
    (df_inconsistent["toplam_tutar"] <= 0) | 
    (df_inconsistent["teslimat_gunu"] < 0) |
    (df_inconsistent["musteri_puani"] > 5)
]
```

```python
inconsistent_rows
```

```python
len(inconsistent_rows)
```

```python
df_inconsistent = df_inconsistent.drop(index= inconsistent_rows.index)
```

##### 13. birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın. Sonra aykırı değer içeren kayıt sayısını yazdırın.

```python
q1 = df_inconsistent["birim_fiyat"].quantile(0.25)
q3 = df_inconsistent["birim_fiyat"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
```

```python
outliers = df_inconsistent[(df_inconsistent["birim_fiyat"] < lower) | (df_inconsistent["birim_fiyat"] > upper)]
```

```python
len(outliers)
```

```python
df_preprocessed = df_inconsistent.copy()
```
